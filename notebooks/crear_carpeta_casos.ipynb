```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import shutil
```

```python
Hs = ['Hs1','Hs2','Hs3','Hs4','Hs5']
Tp = ['Tp1','Tp2','Tp3','Tp4','Tp5']
direction = ['dir1','dir2','dir3','dir4','dir5']
cases = ['1','2']
```

```python
gid = rf'D:\arturo_sim\simulaciones\oc4_0_1.gid'

for i in Hs:
    for j in Tp:
        for k in direction:
            for case in cases:

                destino = rf'D:\arturo_sim\simulaciones\acoplado\listado\oc4__{i}__{j}__{k}__{case}.gid'
                shutil.copytree(gid, destino)
                archivos = os.listdir(destino)
                
                for archivo in archivos:
                    if archivo.startswith('oc4_0_1'):
                        # reemplazo en el filename
                        nuevo_nombre = archivo.replace('oc4_0_1',f'oc4__{i}__{j}__{k}__{case}')

                        # Renombrar el archivo
                        viejo_path = os.path.join(destino, archivo)
                        nuevo_path = os.path.join(destino, nuevo_nombre)

                        os.rename(viejo_path, nuevo_path)
```
